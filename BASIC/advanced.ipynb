### LIST COMPREHENSION

```python
lst = [4,5,7,2,3,9,6]
even = [x for x in lst if x % 2 == 0]
print(even)

cube = [x ** 3 for x in range(10) if x ** 3 > 50]
print(cube)

upper = [x.upper() for x in "hello Guys" if x != ' ']
print(upper)
```

### LAMBDA FUNCTION
Anonymous functions that do not contains any naming convention and are useful in making task easier.

```python
#lambda arguments : expression
cube = lambda x: x ** 3
print(cube(6))
```

### MAP
It is used to perform operations on each element of any iterable(alternative to loop).Takes function name and iterable.

```python
num = [8,4,5,3,2,6,9,2,22]
def even_number(num):
    if num % 2 == 0:
        return num

data = list(map(even_number, num))  #fetches each element. Maps the value to original value.[elements in iterable are intact.]
print(data)


def square(num):
    return num ** 2

print(list(map(square, num)))


words = ["hello", "world", "python"]
lengths = list(map(len, words)) #returns length of each word of words.
print(lengths)
```

### FILTER
It is used to filter the values based on condition in any iterable.

```python
num = [8,4,5,3,2,6,9,2,22]
def even_number(num):
    if num % 2 == 0:
        return num

data = list(filter(even_number, num))  #fetches each element, apply conditions and return only those which passes check.[elements in iterable are removed.]
print(data)
```

### ZIP
It pairs elements of two iterable together.

```python
x = [101, 102, 103, 104, 105]
y = ['abhi', 'akash', 'ani']
data = dict(zip(x,y))
data_list = list(zip(x,y))
print(f'{data} \n{data_list}')
```

### ENUMERATE
It returns a tuple containing pair of index and value.

```python
y = ['abhi', 'akash', 'ani', 'ashi', 'krish','akki']
print(list(enumerate(y)))
print(list(enumerate(y, 101)))    # second parameter controls the index to start.

fruits = ['apple', 'banana']
for index, fruit in enumerate(fruits):
    print(index)  
    print(fruit) 
```

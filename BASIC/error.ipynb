## ERROR HANDLING

### TRY EXCEPT:

```python
a = int(input("enter a number:"))
b = int(input("enter a number:"))

try:
    a / b
except Exception as error:
    print("error found:", error)
```

### ELSE & FINALLY:

```python
a = int(input("enter a number:"))
b = int(input("enter a number:"))

try:
    result = a / b
except Exception as error:
    print("error found:", error)
else:
    print("the result is: ", result)

finally:
    print("always runs")
```

### RAISE:

```python
a = int(input("enter a number:"))
b = int(input("enter a number:"))

if b == 0: 
    raise ValueError("Divisble by Zero")
else:
    print(a/b)
    
```

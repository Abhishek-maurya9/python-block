### TYPES OF ATTRIBUTES AND METHODS
self is a parameter that tracks the location of object that is instantiated using class.It points out to the class instances(objects).

```python
class Animal:
    age = 12    # Class Attribute

    def __init__(self, name, color):
        self.name = name    #Instance Attribute - takes location of objects and are accessible to each objects only.
        self.color = color

    def sound(self, voice): #Instance Method 
        self.voice = voice

    @classmethod    #It points the location of class.
    def movement(cls, movement):
        cls.movement = movement
        try:
            print("Object attributes are not accessible ", {cls.name})  #We cannot access instance attribute from class method as it takes location reference of class intead of Object.
        except Exception as error:
            print(error)
        print("Class attribute is: ", cls.age)

    @staticmethod   #No reference here, It can neither point class nor Object.
    def greet():
        print("hello ")

l1 = Animal("Lion", "Orange")
print(f'Using Instance: {l1.age}',f"Using Class: {Animal.age}")
l1.sound("roar")
print(l1.voice)
l1.movement("walk")
```

### INHERITENCE
Allows a class to inherit properties and behaviour(methods and attributes) of another class.
Constructor of Parent class is run directly if not overidden in child class.

```python
class Animal:
    age = 10
    def __init__(self, name, color):
        self.name = name
        self.color = color

    def sound(self, sound):
        self.sound = sound
        return self.sound
    
#Inherit
class Dog(Animal):
    def info(self):
        sound = super().sound("Bark")
        print(self.name, self.color, sound, self.age)

d1 = Dog("German sepherd", "brown")
d1.info()
```

### POLYMORPHISM
Interface and methods having same name behaves differently depending upon objects.
In python there is no concept of `method overloading` directly but can be acheived by multiple ways.
while trying method overloading means same method name but different parameters, it converted to `method overriding`.

```python
class Animal:
    age = 10
    def __init__(self, name, color):
        self.name = name
        self.color = color

    def sound(self):
        print("Sounds")

    def sound(self, sound):
        print(sound)


class Cat(Animal):
    def __init__(self,breed):   #conceptually method overloading.
        self.breed = breed

    def sound(self):
        print("meow ")

class Horse(Animal):
    def __init__(self,breed, name, color):
        self.breed = breed
        super().__init__(name, color)

try:
    c1 = Cat("toi", "red","indian") # init constructor is overridden therefore name and color cannot be given like this.
except Exception as error:
    print(error)


c2 = Cat("indian")
c2.name = "toui"
print(c2.name)

c2.sound()    # sound now becomes meow(got overidden from Sounds to meow).

#method overloading(but it is overriding)
h1 = Horse("german", "uiop", "grey")
h1.sound("hihihihi")

try:
    h1.sound()
except Exception as err:
    print(err)
```

### ENCAPSULATION
It involves bundling of data(attributes and methods) into a single unit. To protect it from unauthorized access and provides controled modification.
Here comes concept of visibility(Private __ and protected _).

```python
class Bank:
    __total_amount = 1000000
    _age = 78
    def set_total(self, amount):
        self.__total_amount += amount

    def get_total(self):
        return self.__total_amount

b1 = Bank()

try:
    b1.__total_amount
except Exception as er:
    print(er)

print(b1._age)

b1.get_total()
```

### ABSTRACTION
It is same like abstract contract where we need to add function signature or functions.Providing predefined methods to enforce in main class.If interface function is not enforced then throws error.

```python
from abc import ABC, abstractmethod

class enforce(ABC):
    @abstractmethod
    def start():
        pass

    def stop(self):
        self.brake = True

class Car(enforce):
    def __init__(self):
        self.clutch = False
        self.acc = False
        self.brake = False

    def stop(self):
        self.brake = True
        print("Car stopped:")
        
try:
    c1 = Car()
except Exception as err:
    print(err)
```

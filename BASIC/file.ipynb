### Path module

```python
from pathlib import Path
import os
```

## Path methods:

```python
# 1: adding path:-
name = "hello.txt"
path = Path(name)

# 2: rename path name:-
newname = "abhishek.txt"
newpath = Path(newname)
path.rename(newpath)

# 3: deletes the file:-
path.unlink()

# 4: check whether file exist:
newpath.exists()
```

### File Open methods:

```python
# There is manual closing of file.
file = open("new.txt", 'w')

# The file is automatically closed.
with open("new.txt", 'w') as file:
    pass
```

### File Read methods

```python
# returns complete file content at once.
try:
    with open("demo.txt", 'r') as f:
        content = f.read()
        print(content)

    with open("demo.txt") as f:
        print(f.readline())

    with open("demo.txt") as f:
        print(f.readlines())
except Exception as err:
    print(err)
```

### File Write methods

```python
# It appends into the file if file exists.
with open("demo.txt", "a") as f:
  f.write("\nNow the file has more content!")
with open("demo.txt") as f:
  print(f.read())

# It will overwrite the file, creates file if not exists.
with open("demo.txt", "w") as f:
  f.write("Woops! I have deleted the content!")
with open("demo.txt") as f:
  print(f.read())
```

```python
# This will delete the file.
os.remove("demo.txt")
```

### Seek method 
To point out the file content index

```python
path = Path("hello.txt")

with open(path, 'w') as f:
   f.write("hello i am abhishek maurya and i am a python developer.")

with open(path, 'r') as f:
    content1 = f.read()
    content2 = f.read()
    f.seek(0)
    content3 = f.read()

print(f"{content1} \n{content2} \n{content3}")
path.unlink()
```

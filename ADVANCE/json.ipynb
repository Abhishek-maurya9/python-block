## JSON(JavaScript Object Notation)

```python
import json
```

```python
user = {
    "age": 20,
    'name': "abhi",
    "id": 101,
    "skill": ["blockchain", "AI", "ML"],
    "dev": {"ML dev": 5, "AI dev": 6, "QA dev": 7}
}

result = json.dumps(user, indent=2)
print(result)
print(type(result))
```

```python
with open("file.json", 'w') as f:
    result = json.dump(user, f, indent=4)
    print(type(result))
```

```python
with open("file.json", 'r') as f:
    content = f.read()

    result = json.loads(content)
    print(result)
    print(type(result))
    print(result["skill"][0])
    print(result["dev"]["QA dev"])
```

```python
#directly reads json data from file and convert to python object.
with open("file.json", 'r') as f:
    result = json.load(f)
    print(result)
    print(type(result))
    print(result["skill"][0])
    print(result["dev"]["QA dev"])
```

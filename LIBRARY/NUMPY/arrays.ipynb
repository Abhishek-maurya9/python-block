## Numpy Library

```python
import numpy as np
```

### Creating 1D array from list and tuple

```python
arr_1d = np.array([2,3,5,6,7])
print("1D array using list:", arr_1d)

arr__1d = np.array((2,3,5,6,7))
print("1D array using tuple:", arr__1d)
```

### Creating a 2D array
`array()` always takes array(list or tuple) as input

```python
arr_2d = np.array([[2,3,4,5,6],[4,7,8,9,0]])
print("2d array: ", arr_2d)
```

### List vs numpy array

```python
py_list = [8,7,4,3]
print("the list multiplication: ", py_list * 2)

np_array = np.array([8,7,4,3])
print("the array multiplication: ", np_array * 2) # it does element wise multiplication.Scalar Multiplication
```

### Create array using `arange()`
Only creates 1D array, but can be changed to other dimensions using `reshape(rows, column)`.

```python
arr = np.arange(0,20,2) * 2
print(arr)

arr = np.arange(0,20,2).reshape(5,2)
print(arr)
```

### Creating Arrays:
It takes shapes as input in either [] or ().

```python
#Zero array: zeros()
zero_arr = np.zeros((3,4))
print("zero array \n", zero_arr)

#Array of 1: ones()
one_arr = np.ones((4,4))
print("one array \n", one_arr)

#Constant value array: full()
const_arr = np.full((3,4),6)
print("constant array \n", const_arr)

randon_arr = np.random.random((2,3))
print("random values array: \n", randon_arr)

identity_arr = np.eye(4)
print("identity matrix: \n", identity_arr)

empty_arr = np.empty((2,2))
print("empty array: \n", empty_arr)
```

### MULTIDIMENSIONAL ARRAY

```python
arr5 = np.array([1, 2, 3, 4], ndmin=5)  #No. of dimensions = No. of square brackets
print(arr5, arr5.ndim, sep=",")
```

### VECTOR, MATRIX AND TENSOR

```python
vector = np.array([2,3,4,5,6])
print("vector: ", vector)

matrix = np.array([[2,3,4,5],[7,6,5,4]])
print("matrix: ", matrix)

#tensor is just an array of multidimensions.Ex - (It takes 2d array as 1 array.)
tensor = np.array([[[3,4,5],[6,7,8]],
                   [[3,5,6],[8,9,6]]])
print("tensor: ", tensor)
```

### ARRAY PROPERTIES:

```python
arr = np.array([[3,4,5],[6,7,4]])
print("size: ",arr.size)
print("dimension: ", arr.ndim)
print("shape: ", arr.shape)
print("Data Type: ", arr.dtype)
```

### ARRAY METHODS:

```python
arr = np.arange(12)
print("normal array: \n", arr)

reshaped = arr.reshape((3,4))
print("reshaped array: \n", reshaped)

#This will return a copy of original array.
flattened = reshaped.flatten()
print("flattened array: \n", flattened)

#This will return the view of array(or the original array is returned).
raveled = reshaped.ravel()
print("raveled array: \n", raveled)

transpose = reshaped.T
print("Transpose of array: \n", transpose)
```

### ARRAY SLICING:

```python
arr_1D = np.arange((12))
print("sliced array: ", arr_1D[2:12:2])

array_2D = np.array([[2,3,4,5],[6,8,4,3],[4,5,6,7]])
print("sliced array_2d: ", array_2D[1,2]) #return 4
print("sliced array_2d: ", array_2D[:, 2]) # returns 2nd column(: is for complete rows) 
print("sliced array_2d: ", array_2D[:, :]) 
```

### ARRAY SORTING:

```python
arr = np.array([5,4,6,7,4,3,5,2,9,1])
print("sorted array: ", np.sort(arr))

arr_2D = np.array([[2,5,4,3],[6,8,4,3],[4,7,6,5]])
print("sorted array: ", np.sort(arr_2D, axis=0))    #Column wise sorting
print("sorted array: ", np.sort(arr_2D, axis=1))    #Row wise sorting
```

### FILTER and MASK:

```python
arr = np.array([5,4,6,7,4,3,5,2,9,1])
print("Even numbers: ", arr[arr % 2 == 0])
print("Even numbers: ", arr[arr > 4])

#With MASK
mask = arr % 3 == 0
print(mask)
print("Divisible by 3: ", arr[mask])
```

### FANCY INDEXING AND WHERE:

```python
arr = np.array([5,4,6,7,4,3,5,2,9,1])
indices = [0,3,5,6,8]
print("array: ", arr[indices])

arr_where = np.where(arr > 5)
print(arr_where)    #stores indexes of the condition.
print("array: ", arr[arr_where])

## TERNARY OPERATOR:(Array creation)
where_condition = np.where(arr > 5, arr * 3, arr)   #(a>b ? true: false)like this.
print("array: ", where_condition)
```

### ARRAY CONCAT:

```python
arr1 = np.arange(4)
arr2 = np.arange(5,10)
print("combined array: ", np.concatenate((arr1, arr2)))

arr1 = np.array([8,9,7])
arr2 = np.array([6,2,5])
print("sum array: ", arr1 + arr2)
```

### ADDING AND REMOVING ROW/COLUMN
use `vstack` to add row in array and `hstack` to add columns.
Both function takes a tuple as input which includes original array and new array.
use `delete` function to delete the element at given index. It takes original array and index as input. 

```python
original = np.array([[8,9],[5,7]])
new_row = np.array([2,1])

add_row = np.vstack((original, new_row))
print("new array with new row \n", add_row)

new_col = np.array([[6],[0]])
add_column = np.hstack((original, new_col))
print("new array with new columns \n", add_column)


deleted_arr = np.delete(original, 2)
print("new array after deletion: \n", deleted_arr)
```

### VECTORIZATION
Vectorization means performing operations on an entire NumPy array instead of manually iterating through its elements using loops.

```python
numbers = [1, 2, 3, 4, 5]
result = []
for num in numbers:
    result.append(num * 2)

print(result)  #iterating each elements, checking their data types and then performing operations,increases computational cost.

arr = np.arange(10)
print(arr * 2)  #Vectorization (ex- scalar multiplication) performs operation in entire array at once.cost efficient

#More Operations:
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

### BROADCASTING
Broadcasting is NumPy's mechanism for performing operations between arrays of different shapes without explicitly copying or manually expanding the smaller array.

```python
import numpy as np

a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

b = np.array([10, 20, 30])

print(a + b)
```

```python
#VOWELS CHECKER:
def checkVowels(str):
    count = 0;
    for x in str.lower():
        if(x == 'a' or x == 'e' or x == 'i' or x == 'o' or x == 'u'):
            count += 1
    return count

print(checkVowels("Abhishek Maurya Is A Developer"))
```

```python
#SECOND LARGEST:
def secondLargest(lst):
    largest = float('-inf')
    s_largest = float('-inf')
    n = len(lst)

    for i in range(0,n):
        largest = max(largest, lst[i])
    
    for i in range(0,n):
        if(lst[i] > s_largest and lst[i] != largest):
            s_largest = lst[i]
    print(s_largest)
secondLargest([3,5,8,5,3])
```

```python
#MISSING VALUE IN RANGE:
def missingValue(lst):
    n = len(lst) + 1
    num = range(1,n)

    for i in num:
        if(i not in lst):
            return i
print(missingValue([2, 3, 1, 5]))
```

```python
#TWO SUM
x = 13
lst = [5, 9, 1, 2, 4, 10, 6, 3]
n = len(lst)
hash = {}
new_list = []
def getNum(lst):
    for i in range(n):
        remaining = x - lst[i]
        if remaining in hash:
            new_list.extend([(hash[remaining], i)])
        hash[lst[i]] = i

(getNum(lst))
print(new_list)
```

```python
#Armstrong
num = int(input("enter a number"))
n = num
result = 0
while n > 0:
    rem = n % 10
    result = result + rem ** 3
    n = n // 10

if result == num:
    print("armstrong")
else:
    print("not a armstrong")
```

```python
#Perfect Number
num = int(input("enter a number"))
n = num
result = []
sum = 0
for i in range(1,num):
    if n % i == 0:
        result.append(i)

for i in result:
    sum += i

if sum == num:
    print("perfect Number")
```

```python
#Factorial
num = int(input("enter a number"))
result = 1
while num > 0:
    result *= num
    num -= 1
print(result)

def fact(num):
    if(num == 1 or num == 0):
        return 1
    else:
        return fact(num - 1) * num

print(fact(5))
```

```python
#reverse an array:
def reverseArray(arr, l, r):
    if l >= r:
        return
    arr[l],arr[r] = arr[r],arr[l]
    reverseArray(arr, l + 1, r - 1)
    return arr

result = reverseArray([3,6,7,5,2,1,9], 1, 6)
print(result)


#reverse an array:
def reverseArray(str, l, r):
    if l >= r:
        return True
    if str[l] != str[r]:
        return False
    reverseArray(str, l + 1, r - 1)
    return True

str = 'NITIN'
result = reverseArray(str, 0, len(str) - 1)
if result == True:
    print("pallindrom")
```

```python
#Fibonnaci series
def fibbonaci(num):
    if num == 0:
        return 0
    if num == 1:
        return 1
    return fibbonaci(num - 1) + fibbonaci(num - 2)

print(fibbonaci(9))
```

```python
#Largest and second largest value:
def search(arr):
    largest = float('-inf')
    second_largest = float('-inf')

    for i in range(len(arr)):
        if arr[i] > largest:
            second_largest = largest
            largest = arr[i]
        elif arr[i] > second_largest and arr[i] != largest:
            second_largest = arr[i]

    return largest, second_largest

print(search([3,4,3,2,5,9,7]))
```
